## Let's Build Our First Artificial Deep Neural Network (ANN)

## 1. Import Necessary Libraries

```python
import keras
```

## 2. Data Collection

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
```

## 4. Data Preparation

### Let's Normalize this for Faster processing

```python
x_train.min(),x_train.max()
```

```python
x_test.min(),x_test.max()
```

```python
x_train.mean(),x_train.std()
```

```python
x_test.mean(),x_test.std()
```

```python
x_train = x_train/255.0
x_test = x_test/255.0
```

```python
x_train.min(),x_train.max()
```

```python
x_test.min(),x_test.max()
```

## 5. Model Building

```python
from keras.models import Sequential
from keras.layers import Flatten,Dense
```

```python
model = Sequential()
model.add(Flatten())
model.add(Dense(units=100,activation="relu",))
model.add(Dense(units=60,activation="relu",))
model.add(Dense(units=10,activation="softmax"))
#model.summary()
```

```python
model.compile(optimizer='rmsprop',loss='sparse_categorical_crossentropy')
```

## 6. Model Training || 7. Model Testing || 8. Model Evaluation

```python
model.fit(x=x_train,y=y_train,batch_size=32,epochs=10,validation_data=(x_test,y_test))
```

```python
model.summary()
```
